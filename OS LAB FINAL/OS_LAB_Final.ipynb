````python
def fcfs(q,head):
  total = 0
  current=head
  print('path : --->',head,end='')
  for track in q:
    print('\nCurrent Head :',track)
    move=abs(track-current)
    print(f"|{track}-{current}|={move}")
    total+=move
    current=track
  print('total distance :',total)
#q=list(map(int,input("Enter Number a : ").split()))
#head=int(input("Enter Number b : "))
q=[98,183,37,122,14,124,65,67]
h=53
fcfs(q,h)
````

````python
def sstf(q,h):
  total=0
  current=h
  print('path : --->',h,end='')
  while len(q)>0:
    near=q[0]
    min_dist=abs(near-current)
    for i in q:
      dist=abs(i-current)
      if dist<min_dist:
        min_dist=dist
        near=i

    move=abs(near-current)
    print(f"|{near}-{current}|={move}")
    total+=move
    current=near
    q.remove(near)
    print('path : --->',near,end='')
  print('total distance :',total)
q=[98,183,37,122,14,124,65,67]
h=53
sstf(q,h)
````

````python

def cscan(queue, head, direction, disk_size=199):
    total = 0
    current = head

    left = []
    right = []


    for track in queue:
        if track < head:
            left.append(track)
        else:
            right.append(track)

    left.sort()
    right.sort()

    print("Path:", current, end=" ")

    if direction == "inc":

        for track in right:
            movement = abs(current - track)
            print(f"\nMove: |{current}-{track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")


        if current != disk_size:
            movement = abs(current - disk_size)
            print(f"\nMove to max cylinder: |{current}-{disk_size}| = {movement}")
            total += movement
            current = disk_size
            print(disk_size, end=" ")

        total+=disk_size
        print("\nJump from max to 0")
        current = 0
        print(current, end=" ")


        for track in left:
            movement = abs(current - track)
            print(f"\nMove: |{current}-{track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")
    else:

        left.sort(reverse=True)
        for track in left:
            movement = abs(current - track)
            print(f"\nMove: |{current}-{track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")

        if current != 0:
            movement = abs(current - 0)
            print(f"\nMove to 0: |{current}-0| = {movement}")
            total += movement
            current = 0
            print(current, end=" ")

        total+=disk_size
        print("\nJump from 0 to max cylinder")
        current = disk_size
        print(current, end=" ")

        right.sort(reverse=True)
        for track in right:
            movement = abs(current - track)
            print(f"\nMove: |{current}-{track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")
    print("\nTotal head movement =", total)
q=[98,183,37,122,14,124,65,67]
h=53
d='inc'
cscan(q,h,d)
````

````python
def scan(queue, head, direction, disk_size=199):
    total = 0
    current = head

    left = []
    right = []

    for track in queue:
        if track < head:
            left.append(track)
        else:
            right.append(track)

    left.sort(reverse=True)
    right.sort()

    print("Path:", current, end=" ")

    if direction == "dec":   # decreasing order

        for track in left:
            movement = abs(current - track)
            print(f"\nMove: |{current} - {track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")

        if current != 0:
            movement = abs(current - 0)
            print(f"\nMove: |{current} - 0| = {movement}")
            total += movement
            current = 0
            print(0, end=" ")

        for track in right:
            movement = abs(current - track)
            print(f"\nMove: |{current} - {track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")

    else:   # increasing order

        for track in right:
            movement = abs(current - track)
            print(f"\nMove: |{current} - {track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")

        if current != disk_size:
            movement = abs(current - disk_size)
            print(f"\nMove: |{current} - {disk_size}| = {movement}")
            total += movement
            current = disk_size
            print(disk_size, end=" ")

        for track in left:
            movement = abs(current - track)
            print(f"\nMove: |{current} - {track}| = {movement}")
            total += movement
            current = track
            print(track, end=" ")

    print("\nTotal head movement =", total)


queue = [98,183,37,122,14,124,65,67]
head = 53
direction ='inc'
scan(queue, head, direction)
````

1. FCFS (First Come First Serve)

FCFS serves disk requests in the order they arrive.

If requests come like: 40 → 10 → 70 → 20

and head is at 50, it will go exactly in that order.

How it works:

No sorting

Very simple

Just follow queue order

Problem:

Can cause large head movement

Not efficient for big workloads

2. SSTF (Shortest Seek Time First)

SSTF selects the request that is closest to the current head position.

How it works:

From current head, choose nearest request first

Then again choose nearest from new position

Example idea:

If head is at 50 and requests are 10, 20, 60, 70

It will go: 60 (closest), then 70, then 20, then 10

Problem:

Can cause starvation (far requests may wait too long)

3. SCAN

SCAN moves the disk arm in one direction and serves all requests on the way, then reverses.
It works like an elevator going up and down.

How it works:

Head moves left or right

It serves all requests in that direction

When it reaches the end of disk, it changes direction

Then it continues serving remaining requests

4. C-SCAN (Circular SCAN)

C-SCAN moves in one direction only and serves requests, then jumps back to start.
It gives more uniform waiting time than SCAN.

How it works:

Disk head moves in one direction only (example: left → right)

It serves all requests on the way

When it reaches the end of the disk, it jumps back to the start

Then again moves in same direction

Important point:

It serves nothing during the return jump

5. C-LOOK

C-LOOK is like C-SCAN but only goes up to the last request, not the end of the disk.
Then it jumps back to the first request, reducing extra movement.

How it works:

Disk head moves in one direction only

It serves requests only up to the last request (not full disk end)

Then it jumps directly to the first request

Again continues same direction

Important point:

It does NOT go to the extreme ends of the disk

C-LOOK is considered the best because it reduces unnecessary head movement and gives good performance.
It only goes where requests exist, so it is faster and more efficient than SCAN and C-SCAN.

Easy Comparison
1. FCFS

Very simple

But slow because head moves too much

 Worst performance among all

2. SSTF

Reduces head movement

But far requests may wait too long (starvation)

 Faster than FCFS but not fully fair

3. SCAN

Fair and organized

But still goes to disk end even if no request is there

 Better than SSTF

4. C-SCAN

Gives equal waiting time

But jumps from end to start, causing extra movement

 More fair than SCAN

5. C-LOOK

Goes only to the last request, not full disk end

Less movement and faster service

 Best balance of speed and fairness

# Page Replacing Algorithms

First-In-First-Out

````python
def fifo(ref, frames):
    memory = []
    miss = 0
    hit=0
    for page in ref:
        if page not in memory:
            if len(memory) < frames:
                memory.append(page)
            else:
                memory.pop(0)
                memory.append(page)
            miss += 1
        else:
          hit +=1
        print(page, "->", memory)

    print('\nTotal Miss =',miss)
    print('\n Total Hit =',hit)



pages = [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]
frames = 3

fifo(pages, frames)
````

Least Recently Used (LRU) Algorithm

````python
def lru(ref, frames):
    memory = []
    miss = 0
    hit = 0

    for page in ref:

        if page in memory:
            hit += 1
            memory.remove(page)
            memory.append(page)
        else:
            miss += 1

            if len(memory) < frames:
                memory.append(page)
            else:
                memory.pop(0)
                memory.append(page)

        print(page, "->", memory)

    print("Miss =", miss)
    print("Hit =", hit)
pages = [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]
frames = 3
lru(pages,frames)
````

Optimal Algorithm

````python
def optimal(pages, frames):

    memory = []
    miss = 0
    hits = 0

    for i in range(len(pages)):
        page = pages[i]

        if page in memory:
            hits += 1
            fault = "hit"

        else:

            miss += 1
            fault = "miss"

            if len(memory) < frames:
                memory.append(page)

            else:
                future = pages[i+1:]

                index_to_replace = -1
                farthest = -1

                for j in range(len(memory)):

                    if memory[j] not in future:
                        index_to_replace = j
                        break

                    else:
                        pos = future.index(memory[j])

                        if pos > farthest:
                            farthest = pos
                            index_to_replace = j

                memory[index_to_replace] = page

        print(f"Page: {page} -> Frames: {memory} -> {fault}")

    total = len(pages)
    hit_ratio = hits / total
    miss_ratio = miss / total

    print("\nTotal Pages:", total)
    print("Total Hits:", hits)
    print("Total Misses:", miss)
    print("Hit Ratio:", round(hit_ratio, 2))
    print("Miss Ratio:", round(miss_ratio, 2))


pages = [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]
frames = 3

optimal(pages, frames)
````

1. FIFO (First In First Out)

Idea:
The page that came first into memory will be removed first.

How it works:

Maintain a queue.

Oldest page (first inserted) is replaced first.

Example:

If pages are: A, B, C, D

Then A goes out first when replacement is needed.

Pros:

Very easy to implement

Cons:

May remove frequently used pages

Can give bad performance (Belady’s anomaly)

2. LRU (Least Recently Used)

Idea:
Remove the page that was not used for the longest time.


How it works:

Track page usage history.

When replacement needed, remove least recently accessed page.

Example:

If A used long ago and B used recently → A will be removed.

Pros:

Good performance in real systems

Uses past behavior to predict future

Cons:

Hard to implement (needs tracking/time stamps or stack)

3. Optimal Page Replacement

Idea:
Remove the page that will not be used for the longest time in the future.

How it works:

Look into future page requests.

Replace page that is needed farthest in future.

Example:

If page A will be used after 100 steps and B after 5 steps → remove A.

Pros:

Best possible performance

Minimum page faults

Cons:

Impossible in real OS (future prediction needed)

# CPU Time Scheduling Algorithms

First Come First Serve

````python
def fcfs(processes, bt):
    n = len(processes)
    wt = [0] * n
    tat = [0] * n

    tat[0] = bt[0]

    for i in range(1, n):
        wt[i] = wt[i - 1] + bt[i - 1]
        tat[i] = wt[i] + bt[i]

    print("Process  BT  WT  TAT")
    for i in range(n):
        print(processes[i], bt[i], wt[i], tat[i])

    avg_wt = sum(wt) / n
    avg_tat = sum(tat) / n

    print("Average Waiting Time:", avg_wt)
    print("Average Turnaround Time:", avg_tat)
pro=[1,2,3,4]
br=[12,3,22,25]
fcfs(pro,br)
````

Shortest Job First

````python
def sjf(processes, bt):
    n = len(processes)

    for i in range(n):
        for j in range(i + 1, n):
            if bt[i] > bt[j]:
                bt[i], bt[j] = bt[j], bt[i]
                processes[i], processes[j] = processes[j], processes[i]

    wt = [0] * n
    tat = [0] * n

    for i in range(1, n):
        wt[i] = wt[i-1] + bt[i-1]

    for i in range(n):
        tat[i] = wt[i] + bt[i]

    print("Process  BT  WT  TAT")
    for i in range(n):
        print(processes[i], "     ", bt[i], " ", wt[i], " ", tat[i])

    avg_wt = sum(wt) / n
    avg_tat = sum(tat) / n

    print("Average Waiting Time:", avg_wt)
    print("Average Turnaround Time:", avg_tat)
pro=[1,2,3,4]
br=[12,3,22,25]
sjf(pro,br)
````

Priority Scheduling

````python
def priority_scheduling(processes, bt, priority):
    n = len(processes)
    wt = [0] * n
    tat = [0] * n


    for i in range(n):
        for j in range(i + 1, n):
            if priority[i] > priority[j]:

                priority[i], priority[j] = priority[j], priority[i]

                bt[i], bt[j] = bt[j], bt[i]

                processes[i], processes[j] = processes[j], processes[i]

    tat[0] = bt[0]

    for i in range(1, n):
        wt[i] = wt[i - 1] + bt[i - 1]
        tat[i] = wt[i] + bt[i]

    print("Process  BT  Priority  WT  TAT")
    for i in range(n):
        print(processes[i], bt[i], priority[i], wt[i], tat[i])

    print("Average Waiting Time:", sum(wt)/n)
    print("Average Turnaround Time:", sum(tat)/n)


p=[1,2,3,4]
b=[12,3,22,25]
pr=[2,1,4,3]

priority_scheduling(p, b, pr)
````

Round Robin

````python
def round_robin(p, bt, tq):
    n = len(p)
    rem = bt.copy()
    wt = [0]*n
    time = 0
    e=[]
    e2=[]

    while sum(rem) > 0:
        for i in range(n):
            if rem[i] > 0:
                if rem[i] <= tq:
                    time += rem[i]
                    wt[i] = time - bt[i]
                    rem[i] = 0
                    e2.append(p[i])
                    e.append(time)
                else:
                    time += tq
                    rem[i] -= tq
                    e2.append(p[i])
                    e.append(time)

    print("Process  BT  WT  TAT")
    for i in range(n):
        tat = wt[i] + bt[i]
        print(p[i], bt[i], wt[i], tat)
    for i in range(len(e)):
        print(f'process {e2[i]} TAT {e[i]}')


    print("Average WT =", sum(wt)/n)
    print("Average TAT =", sum(wt[i] + bt[i] for i in range(n))/n)



p=[1,2,3,4]
bt=[12,3,22,25]
tq = 10

round_robin(p, bt, tq)
````

SJF

Priority Scheduling

# 1. What is a System Call?

A system call is a mechanism that allows a user program (running in user mode) to request a service from the operating system (running in kernel mode). It acts as an interface between user programs and the OS, enabling safe access to hardware and system resources that user programs cannot access directly.

## Examples of System Calls
- read()
- write()
- open()
- close()
- fork()
- exec()

## How it Works

### 1. User Process Executing (User Mode)
The user program runs normally in user mode. It cannot directly use hardware or important system parts and depends on the operating system to do these safely.

### 2. Gets System Call (User Mode)
When the program needs help from the operating system, like reading a file, it makes a system call. This request is still done in user mode.

### 3. Execute System Call (Kernel Mode)
The CPU switches to kernel mode to carry out the request safely. In this mode, the operating system can use the hardware and memory to complete the task.

### 4. Return from System Call (User Mode)
After the system call is done, control goes back to the user program. The CPU switches to user mode again, and the program continues with the results from the system call.

# 2. What is fork() System Call?

The `fork()` system call is used to create a new process called a child process by duplicating the parent process.

## Process of fork()

When a program reaches the `fork()` system call during execution, a new child process is created. The child process receives a copy of the parent’s memory image, including code, data, and stack. Then both parent and child continue execution from the next instruction after `fork()`.

### Return Values
- Child process → 0
- Parent process → Child PID
- Error → -1

Finally, both parent and child processes execute independently.

## Purpose of fork()

### 1. Process Creation
It allows a running program to create a new process.

### 2. Parallelism and Concurrency
Parent and child processes can run at the same time.

### 3. Distinguishing Processes
The return value helps identify parent and child processes.

### 4. Separate Memory Space
The child gets a duplicate copy of the parent’s memory space.

### 5. Common Use Cases
- Shell command execution
- Network servers
- Parallel computing

# 3. What is exec() System Call?

The `exec()` system call is used to replace the current process’s memory image (code, data, and stack) with a new program. It allows a process to execute a completely different executable.

## Process of exec()

When a process calls `exec()`, the operating system removes the existing code, data, and stack of the process and loads a new executable into the same process.

After that:
- Execution starts from the beginning of the new program.
- The Process ID (PID) remains the same.
- The program running inside the process is completely changed.

If `exec()` is successful, it does not return to the old program. It only returns an error if the new program cannot be loaded.

## Purpose of exec()

### 1. Executing New Programs
It loads and runs a new program inside the current process.

### 2. Command Execution in Shells
Shells use `exec()` to run commands such as `ls` or `gcc`.

### 3. Replacing Current Process
It completely transforms the current process into another program.

### 4. Works with fork()
Usually, after `fork()` creates a child process, `exec()` is used to load the new program into that child process.

# 4. What is the Main Difference Between fork() and exec()?

The main difference is that `fork()` creates a new process, while `exec()` replaces the current process with a new program.

## Key Differences Between fork() and exec()

| Feature | fork() | exec() |
|---|---|---|
| Primary Purpose | Duplicates the existing process | Replaces the current process with a new program |
| Process ID (PID) | Creates a new PID for child process | Keeps the same PID |
| Process Count | Increases process count by one | Process count remains same |
| Memory Image | Copies parent’s memory | Replaces memory with new program |
| Execution Path | Parent and child continue from same point | Starts from beginning of new program |
| Return Behavior | Returns in both parent and child | Does not return unless failed |

# 2. What is fork() system call? Write its definition and purpose.
**Ans:** `fork()` is a system call that creates a new process called a child process by duplicating the parent process.

**Process of fork():**

As the program runs step by step, when execution reaches the `fork()` system call, a new child process is created. The child process receives a copy of the parent's memory image, including code, data, and stack. It is then added to the operating system's process list and scheduled for execution. After that, both parent and child processes continue execution from the next instruction following `fork()`. They are distinguished by different return values:
- **Child process** → 0
- **Parent process** → child's process ID (PID)
- **Error (if child cannot be created)** → -1

Finally, both parent and child processes execute independently of each other.

**Purpose of fork():**
- Create a new child process
- Allow parent and child processes to run simultaneously
- Support multitasking in the operating system
- Run multiple tasks independently
- Help servers and applications handle multiple users or jobs
- Create separate processes for background tasks
- Enable process management and parallel execution

# 3. What is exec() system call? Write its definition and purpose.
**Ans:** `exec()` is a system call that replaces the current process's memory image (code, data, stack) with a new program, allowing the process to execute a completely different executable.

**Process of exec():**

When a process calls `exec()`, it replaces its current program with a new program. The operating system removes the existing code, data, and stack of the process and loads a new executable into the same process. After that, execution starts from the beginning of the new program. The process ID remains the same, but the program running inside it is completely changed. If `exec()` is successful, it does not return to the old program, so any code written after it will not execute. It only returns an error if the new program cannot be loaded. Variants of `exec()` also allow passing command-line arguments to the new program.

**Purpose of exec():**
- Run a program without creating an extra process
- Replace the current terminal session with another program
- Change input/output behavior of the current shell
- Improve script efficiency by avoiding unnecessary process creation
